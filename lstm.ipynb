```python
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional, RepeatVector, TimeDistributed, Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

df = pd.read_csv("/content/drive/MyDrive/fitness_lstm_dataset.csv")
df.head()
```

```python
heart_rate = df["heart_rate_bpm"].values
steps = df["steps"].values
calories = df["calories_burned"].values

print(heart_rate[:10])
```

```python
#Basic / Vanilla LSTM
# Prepare data: 3 previous values -> next value
window = 3

X = []
y = []
for i in range(len(heart_rate) - window):
    X.append(heart_rate[i:i+window])
    y.append(heart_rate[i+window])

X = np.array(X).reshape((len(X), window, 1))
y = np.array(y)

print(X.shape, y.shape)
```

```python
model = Sequential([
    LSTM(50, input_shape=(3, 1)),
    Dense(1)
])
model.compile(optimizer="adam", loss="mse")
model.summary()
```

```python
model.fit(X, y, epochs=20, verbose=0)

new_data = heart_rate[-3:].reshape((1, 3, 1))
prediction = model.predict(new_data, verbose=0)

print("Predicted next value:", prediction[0][0])
```

```python
#Stacked LSTM
window = 20

X = []
y = []
for i in range(len(heart_rate) - window):
    X.append(heart_rate[i:i+window])
    y.append(heart_rate[i+window])

X = np.array(X).reshape((len(X), window, 1))
y = np.array(y)

print(X.shape, y.shape)
```

```python
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(20, 1)),
    LSTM(32),
    Dense(1)
])
model.compile(optimizer="adam", loss="mse")
model.summary()
```

```python
model.fit(X, y, epochs=20, verbose=0)

new_data = heart_rate[-20:].reshape((1, 20, 1))
prediction = model.predict(new_data, verbose=0)

print("Predicted next value:", prediction[0][0])
```

```python
#Bidirectional Lstm
window = 20

X = []
y = []
for i in range(len(heart_rate) - window):
    X.append(heart_rate[i:i+window])
    y.append(heart_rate[i+window])

X = np.array(X).reshape((len(X), window, 1))
y = np.array(y)

print(X.shape, y.shape)
```

```python
#Maltivariate LSTM
window = 20

X = []
y = []
for i in range(len(heart_rate) - window):
    seq = []
    for j in range(window):
        seq.append([heart_rate[i+j], steps[i+j], calories[i+j]])
    X.append(seq)
    y.append(heart_rate[i+window])

X = np.array(X)
y = np.array(y)

print(X.shape, y.shape)
```

```python
model = Sequential([
    LSTM(64, input_shape=(20, 3)),
    Dense(1)
])
model.compile(optimizer="adam", loss="mse")
model.summary()
```

```python
model.fit(X, y, epochs=20, verbose=0)

new_data = X[-1].reshape((1, 20, 3))
prediction = model.predict(new_data, verbose=0)

print("Predicted next heart rate:", prediction[0][0])
```

```python
#Many-to-Many LSTM
window = 5
horizon = 5

X = []
y = []
for i in range(len(heart_rate) - window - horizon):
    X.append(heart_rate[i:i+window])
    y.append(heart_rate[i+window:i+window+horizon])

X = np.array(X).reshape((len(X), window, 1))
y = np.array(y).reshape((len(y), horizon, 1))

print(X.shape, y.shape)
```

```python
#LSTM Autoencoder
window = 20

X = []
for i in range(len(heart_rate) - window):
    X.append(heart_rate[i:i+window])

X = np.array(X).reshape((len(X), window, 1))
y = X  # the target is the same sequence we put in

print(X.shape, y.shape)
```

```python
model = Sequential([
    # Encoder
    LSTM(64, input_shape=(20, 1)),
    # Convert one encoded vector back to 20 time steps
    RepeatVector(20),
    # Decoder
    LSTM(64, return_sequences=True),
    # Produce one value at each time step
    TimeDistributed(Dense(1))
])
model.compile(optimizer="adam", loss="mse")
model.summary()
```

```python
#Text / Next-Word LSTM
with open("/content/fitness_text_corpus.txt") as f:
    sentences = [line.strip() for line in f if line.strip()]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary size:", vocab_size)
```

```python
# Build training pairs: input sequence -> next word
input_sequences = []
for sentence in sentences:
    tokens = tokenizer.texts_to_sequences([sentence])[0]
    for i in range(1, len(tokens)):
        input_sequences.append(tokens[:i+1])

max_len = max(len(seq) for seq in input_sequences)
padded = pad_sequences(input_sequences, maxlen=max_len, padding="pre")

X = padded[:, :-1]
y = padded[:, -1]

print(X.shape, y.shape)
```

```python
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=64),
    LSTM(64),
    Dense(vocab_size, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy"
)
model.summary()
```

```python
model.fit(X, y, epochs=100, verbose=0)

seed_text = "my heart rate is"
tokens = tokenizer.texts_to_sequences([seed_text])[0]
tokens = pad_sequences([tokens], maxlen=max_len - 1, padding="pre")

predicted_id = np.argmax(model.predict(tokens, verbose=0))
next_word = tokenizer.index_word.get(predicted_id, "")

print(seed_text, "->", next_word)
```
