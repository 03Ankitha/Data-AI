```python
text="this is a sample sentence for tokenization demo."
words_tokens = text.split()
print("Word Tokens:" , words_tokens)
char_tokens=list(text)
print("Character Tokens", char_tokens)
```

```python
normalized_text=text.lower()
normalized_text="".join([char for char in normalized_text if char.isalnum() or char.isspace()])
normalized_text="".join(normalized_text.split())
print("original text",text)
print("normalized text",text)
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
import re
text=input("enter any text:")
words = re.sub(r'[^\W\S]','',text).split()
filterd_words=[word for word in words if word not in stop_words]
```

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('punkt')
nltk.download('wordnet')
text='The cats were running and studying'
stemmer=PorterStemmer()
stemmed_works = [stemmer.stem(word) for word in text.split()]
print(stemmed_works)
```
