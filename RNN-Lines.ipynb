```python
window_size = 3

# X = inputs/questions
# y = correct answers

X = []
y = []

for sentence in sequence:
    words = sentence.split()

    for i in range(len(words) - window_size):
        X.append(words[i:i + window_size])
        y.append(words[i + window_size])

print("First 10 input examples:")
print(X[:10])

print("First 10 answers:")
print(y[:10])
```

```python
# Create a list of all unique words
vocabulary = sorted(set(
    word
    for sentence in sequence
    for word in sentence.split()
))

# Give each word a number
word_to_int = {
    word: i for i, word in enumerate(vocabulary)
}

print("Vocabulary size:", len(vocabulary))
print("Word to number:")
print(word_to_int)
```

```python
X_numeric = []
y_numeric = []

for inputs, answer in zip(X, y):
    X_numeric.append([word_to_int[word] for word in inputs])
    y_numeric.append(word_to_int[answer])

X_numeric = np.array(X_numeric)
y_numeric = np.array(y_numeric)

print("X numeric:")
print(X_numeric[:10])

print("y numeric:")
print(y_numeric[:10])
```

```python
X_numeric = X_numeric.reshape(
    (X_numeric.shape[0], X_numeric.shape[1], 1)
)

print("New X shape:", X_numeric.shape)
```

```python
model = Sequential()

model.add(
    SimpleRNN(
        50,
        activation='relu',
        input_shape=(window_size, 1)
    )
)

model.add(Dense(len(vocabulary), activation='softmax'))

print("Model created successfully!")
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy'
)

model.fit(
    X_numeric,
    y_numeric,
    epochs=500,
    verbose=0
)

print("Training complete.")
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense


sentence = "I love to play cricket"


words = sentence.split()


vocabulary = sorted(set(words))

word_to_int = {word: i for i, word in enumerate(vocabulary)}
int_to_word = {i: word for word, i in word_to_int.items()}


window_size = 3
X = []
y = []

for i in range(len(words) - window_size):
    X.append([
        word_to_int[words[i + 1]],
        word_to_int[words[i + 2]],
        word_to_int[words[i + 3]]
    ])

    y.append(word_to_int[words[i]])


X = np.array(X)
y = np.array(y)

print("X:", X)
print("y:", y)

X = X.reshape((X.shape[0], X.shape[1], 1))

print("New X shape:", X.shape)

model = Sequential()

model.add(
    SimpleRNN(
        50,
        activation='relu',
        input_shape=(window_size, 1)
    )
)

model.add(Dense(len(vocabulary), activation='softmax'))


model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy'
)


model.fit(X, y, epochs=500, verbose=0)

print("Training complete.")


test_input = ["to", "play", "cricket"]

test_numeric = np.array([
    word_to_int[word] for word in test_input
])

test_numeric = test_numeric.reshape((1, window_size, 1))

predicted = model.predict(test_numeric, verbose=0)

predicted_index = np.argmax(predicted)

predicted_word = int_to_word[predicted_index]

print("Input:", test_input)
print("Predicted previous word:", predicted_word)
```
